This notebook analyzes the Spotify audio dataset containing track audio features (danceability, energy, tempo, valence, etc.) across user playlists.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../datasets/spotify_playlists.csv')
print("Dataset Shape:", df.shape)
df.head()
```

```python
# Summary Statistics
print("Missing Values:\n", df.isnull().sum())
print("\nPlaylist Distribution:\n", df['playlist'].value_counts())
df.describe()
```

```python
# Audio Feature Correlations
audio_cols = ['danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness', 'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo']
plt.figure(figsize=(10, 8))
sns.heatmap(df[audio_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Audio Feature Correlation Matrix')
plt.show()
```
